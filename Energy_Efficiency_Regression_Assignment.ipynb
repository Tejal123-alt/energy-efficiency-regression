```python
# STEP 1: DATA PREPARATION AND INSPECTION

# 1. Dataset information
print("DATASET INFORMATION")
df.info()

# 2. Statistical summary
print("\nSTATISTICAL SUMMARY")
display(df.describe())

# 3. Missing-value check
print("\nMISSING VALUES")
print(df.isnull().sum())

# 4. Duplicate-row check
print("\nDUPLICATE ROWS")
print(df.duplicated().sum())

# 5. Target-variable analysis
print("\nTARGET VARIABLE: Y1 (HEATING LOAD)")
print("Mean:", df["Y1"].mean())
print("Minimum:", df["Y1"].min())
print("Maximum:", df["Y1"].max())

# 6. Categorical-feature inspection
print("\nUNIQUE VALUES IN X6")
print(sorted(df["X6"].unique()))

print("\nUNIQUE VALUES IN X8")
print(sorted(df["X8"].unique()))
```

```python
# STEP 1: DATA CLEANING AND PREPROCESSING

# 1. Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# 2. Check categorical values
print("X6 unique values:", sorted(df["X6"].unique()))
print("X8 unique values:", sorted(df["X8"].unique()))

# 3. Select features and target
X = df.drop(columns=["Y1", "Y2"])
y = df["Y1"]

print("\nFeature shape:", X.shape)
print("Target shape:", y.shape)

# 4. Separate numerical and categorical features
numerical_features = ["X1", "X2", "X3", "X4", "X5", "X7"]
categorical_features = ["X6", "X8"]

print("\nNumerical features:", numerical_features)
print("Categorical features:", categorical_features)
```

```python
# STEP 1: FINAL PREPROCESSING AND TRAIN-TEST SPLIT

from sklearn.model_selection import train_test_split

# Separate features and target
X = df.drop(columns=["Y1", "Y2"])
y = df["Y1"]

# Identify feature types
numerical_features = ["X1", "X2", "X3", "X4", "X5", "X7"]
categorical_features = ["X6", "X8"]

# Split dataset: 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
# STEP 2: MODEL DESIGN AND IMPLEMENTATION

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression

# Preprocessing for multivariate regression
preprocessor = ColumnTransformer(
    transformers=[
        ("num", SimpleImputer(strategy="median"), numerical_features),
        ("cat", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(handle_unknown="ignore"))
        ]), categorical_features)
    ]
)

# MODEL 1: Simple Linear Regression
simple_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("regression", LinearRegression())
])

# MODEL 2: Polynomial Regression (degree 2)
polynomial_model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("polynomial", PolynomialFeatures(degree=2, include_bias=False)),
    ("regression", LinearRegression())
])

# MODEL 3: Multivariate Linear Regression
multivariate_model = Pipeline([
    ("preprocessing", preprocessor),
    ("regression", LinearRegression())
])

print("All three regression models are defined successfully.")
```

```python
# STEP 3: MODEL TRAINING AND EVALUATION

from sklearn.metrics import mean_squared_error, r2_score
import pandas as pd
import numpy as np

# Model 1: use X2 only
simple_model.fit(X_train[["X2"]], y_train)
simple_predictions = simple_model.predict(X_test[["X2"]])

# Model 2: use X2 with polynomial features
polynomial_model.fit(X_train[["X2"]], y_train)
polynomial_predictions = polynomial_model.predict(X_test[["X2"]])

# Model 3: use all selected features
multivariate_model.fit(X_train, y_train)
multivariate_predictions = multivariate_model.predict(X_test)

# Calculate metrics
def evaluate_model(actual, predicted):
    mse = mean_squared_error(actual, predicted)
    rmse = np.sqrt(mse)
    r2 = r2_score(actual, predicted)
    return mse, rmse, r2

results = []

for name, predictions in [
    ("Simple Linear Regression", simple_predictions),
    ("Polynomial Regression", polynomial_predictions),
    ("Multivariate Linear Regression", multivariate_predictions)
]:
    mse, rmse, r2 = evaluate_model(y_test, predictions)
    results.append({
        "Model": name,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    })

results_df = pd.DataFrame(results)
display(results_df.round(4))
```

```python
# STEP 4: COMPARATIVE ANALYSIS

import matplotlib.pyplot as plt

# Compare MSE and RMSE
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].bar(results_df["Model"], results_df["MSE"])
axes[0].set_title("Comparison of MSE")
axes[0].set_ylabel("Mean Squared Error")
axes[0].tick_params(axis="x", rotation=25)

axes[1].bar(results_df["Model"], results_df["RMSE"])
axes[1].set_title("Comparison of RMSE")
axes[1].set_ylabel("Root Mean Squared Error")
axes[1].tick_params(axis="x", rotation=25)

plt.tight_layout()
plt.show()

# Compare R2 scores
plt.figure(figsize=(8, 5))
plt.bar(results_df["Model"], results_df["R2 Score"])
plt.ylabel("R² Score")
plt.title("Comparison of R² Scores")
plt.xticks(rotation=20)
plt.ylim(0, 1)
plt.tight_layout()
plt.show()
```

```python
# Actual vs Predicted values

predictions_dict = {
    "Simple Linear Regression": simple_predictions,
    "Polynomial Regression": polynomial_predictions,
    "Multivariate Linear Regression": multivariate_predictions
}

plt.figure(figsize=(8, 6))

for name, predictions in predictions_dict.items():
    plt.scatter(y_test, predictions, label=name, alpha=0.6)

minimum = min(y_test.min(), min(p.min() for p in predictions_dict.values()))
maximum = max(y_test.max(), max(p.max() for p in predictions_dict.values()))

plt.plot([minimum, maximum], [minimum, maximum], "k--",
         label="Perfect prediction")

plt.xlabel("Actual Heating Load")
plt.ylabel("Predicted Heating Load")
plt.title("Actual vs Predicted Heating Load")
plt.legend()
plt.tight_layout()
plt.show()
```
